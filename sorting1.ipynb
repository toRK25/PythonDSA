```python
def sort(list):
    new_list = []
    while list:
        min_val = max(list)
        new_list.append(min_val)
        list.remove(min_val)
    return new_list
```

```python
list = [5, 2, 9, 1, 5, 6]
```

```python
sort(list)
```

```python
def sort_dsce(arr):
    n = len(arr)
    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            if arr[j] > arr[max_idx]:
                max_idx = j
        arr[i], arr[max_idx] = arr[max_idx], arr[i]
    return arr
```

```python
sort_dsce(list)
```
