DO LIST BNAI , EK fm and EK fn,DO digit ke divisors dono list me story kiye.

phir ek aur list bnai Cf usme jo common divissor hai dono list me vo add kr diya then jo highest value aai vo hoga answer.

```python
def gcd(m,n):
    fm=[]
    for i in range(1,m+1):
        if m%i==0:
            fm.append(i)
    
    fn = []
    for j in range(1,n+1):
        if n%j==0:
            fn.append(j)

    cf= []
    for k in fm:
        if k in fn:
            cf.append(k)
    return max(cf)
```

```python
gcd(12,5)
```

#### can we do it better??

There is a pattern in finding Greatest common divisor is that the divisor should not greater than the minimum of inputs.

ex- GDC(49,23)-the answer can't be greater than 23.

so we can improve our code by using min(m,n)


```python
def gcd(m,n):

    cf=[]
    for i in range(1,min(m,n)+1):
        if m%i==0 and n%i==0:
            cf.append(i)
    return max(cf)
```

```python
gcd(48,18)
```

WE CAN USE EUCLID ALGORITHM

```python
def gcd1(m,n):
    while n != 0:
        m, n = n, m % n
    return m
```
